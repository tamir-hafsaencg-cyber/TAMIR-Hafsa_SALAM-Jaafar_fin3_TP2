```python
import os
import sqlite3
import pandas as pd

# 1. Charger et nettoyer le fichier Excel
file_path = '/content/Tamir Hafsa-Salam jaafar-FIN3-DATA (1).xlsx'
df = pd.read_excel(file_path)

# Nettoyage des colonnes et des données
df.columns = ["facility", "date_of_arrangement", "expiration_date",
              "amount_agreed", "amount_drawn", "amount_outstanding"]

df = df[df["facility"].str.strip().str.lower() != "total"].copy()
df["facility"] = df["facility"].str.strip()

for col in ["date_of_arrangement", "expiration_date"]:
    df[col] = pd.to_datetime(df[col], format="%b %d, %Y").dt.strftime("%Y-%m-%d")

for col in ["amount_agreed", "amount_drawn", "amount_outstanding"]:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(",", ""))

# 2. Création de la base SQL en mémoire et génération du script
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE facilities (
    id                  INTEGER PRIMARY KEY AUTOINCREMENT,
    facility            TEXT NOT NULL,
    date_of_arrangement DATE NOT NULL,
    expiration_date     DATE NOT NULL,
    amount_agreed       REAL,
    amount_drawn        REAL,
    amount_outstanding  REAL
);
""")

df.to_sql('facilities', conn, if_exists='append', index=False)
conn.commit()

# 3. Affichage du script SQL généré
print("-- SCRIPT SQL GÉNÉRÉ --\n")
for line in conn.iterdump():
    print(line)

conn.close()
```

```python
# SECTION 3 : Charger le fichier Excel
# Utilité : remplace le téléchargement Kaggle du tutoriel. Si le fichier n'est pas encore dans
# le dossier, un bouton d'envoi apparaît : choisis ton fichier Excel. Ensuite pandas le lit
# et le met dans un tableau appelé "df".
nom_fichier = "Tamir_Hafsa-Salam_jaafar-FIN3-DATA.xlsx"
excel_path = os.path.join(drive_data_dir, nom_fichier)

if not os.path.exists(excel_path):
    envoye = files.upload()
    shutil.move(list(envoye.keys())[0], excel_path)

df = pd.read_excel(excel_path)
print(f"✅ Excel chargé : {len(df)} lignes, {len(df.columns)} colonnes\n")
display(df.head())
```

```python
# SECTION 4 : Nettoyer les données
# Utilité : prépare le tableau pour SQL.
#  1. renomme les colonnes (sans espaces ni "4/")
#  2. supprime la ligne "Total" (ce n'est pas un vrai enregistrement)
#  3. enlève les espaces inutiles
#  4. convertit les dates "Apr 02, 2025" en format SQL "2025-04-02"
#  5. convertit les montants "3,450,000" (texte) en vrais nombres
df.columns = ["facility", "date_of_arrangement", "expiration_date",
              "amount_agreed", "amount_drawn", "amount_outstanding"]

df = df[df["facility"].str.strip().str.lower() != "total"].copy()
df["facility"] = df["facility"].str.strip()

for col in ["date_of_arrangement", "expiration_date"]:
    df[col] = pd.to_datetime(df[col], format="%b %d, %Y").dt.strftime("%Y-%m-%d")

for col in ["amount_agreed", "amount_drawn", "amount_outstanding"]:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(",", ""))

print(f"✅ Données nettoyées : {len(df)} lignes")
display(df.head())
```

```python
# SECTION 5 : Créer la base de données SQLite et la table
# Utilité : crée la base "imf_facilities.db", définit la structure de la table "facilities"
# (noms et types des colonnes, id automatique), puis y insère toutes les lignes du tableau.
db_path = os.path.join(drive_data_dir, 'imf_facilities.db')
table_name = 'facilities'

if os.path.exists(db_path):          # repartir d'une base vide
    os.remove(db_path)

conn = sqlite3.connect(db_path)
cursor = conn.cursor()

cursor.execute(f"""
CREATE TABLE {table_name} (
    id                  INTEGER PRIMARY KEY AUTOINCREMENT,
    facility            TEXT NOT NULL,
    date_of_arrangement DATE NOT NULL,
    expiration_date     DATE NOT NULL,
    amount_agreed       REAL,
    amount_drawn        REAL,
    amount_outstanding  REAL
)
""")

df.to_sql(table_name, conn, if_exists='append', index=False)
conn.commit()
print(f"✅ Base créée : {db_path}")
print(f"✅ Table '{table_name}' remplie avec {len(df)} lignes")
```

```python
# SECTION 6 : Décrire la structure de la base de données
# Utilité : affiche les colonnes de la table (nom, type), le nombre de lignes,
# et pour chaque colonne le nombre de valeurs uniques et de valeurs vides (NULL).
cursor.execute(f"PRAGMA table_info({table_name})")
columns_info = cursor.fetchall()

print("📋 STRUCTURE DE LA TABLE")
print("=" * 60)
display(pd.DataFrame(columns_info, columns=['cid', 'name', 'type', 'notnull', 'dflt_value', 'pk'])[['name', 'type', 'notnull', 'pk']])

cursor.execute(f"SELECT COUNT(*) FROM {table_name}")
print(f"\nTotal des lignes : {cursor.fetchone()[0]}")

print("\n📈 ANALYSE PAR COLONNE")
print("=" * 60)
for col_info in columns_info:
    col_name, col_type = col_info[1], col_info[2]
    cursor.execute(f"SELECT COUNT(DISTINCT {col_name}) FROM {table_name} WHERE {col_name} IS NOT NULL")
    distincts = cursor.fetchone()[0]
    cursor.execute(f"SELECT COUNT(*) FROM {table_name} WHERE {col_name} IS NULL")
    nulls = cursor.fetchone()[0]
    print(f"{col_name} ({col_type}) -> valeurs uniques : {distincts} | NULL : {nulls}")
```

```python
# SECTION 7 : Requêtes SELECT (vérifier que la base SQL fonctionne)
# Utilité : lit les données avec de vraies requêtes SQL : (1) tout afficher,
# (2) filtrer les facilités qui ont un montant restant à payer, (3) compter et additionner par type de facilité.
print("1️⃣  SELECT simple")
display(pd.read_sql_query(f"SELECT * FROM {table_name}", conn))

print("2️⃣  SELECT avec filtre : montant restant > 0")
display(pd.read_sql_query(f"SELECT * FROM {table_name} WHERE amount_outstanding > 0", conn))

print("3️⃣  SELECT avec agrégation : par type de facilité")
display(pd.read_sql_query(f"""
SELECT facility,
       COUNT(*)                  AS nb_arrangements,
       ROUND(SUM(amount_agreed), 2) AS total_agreed
FROM {table_name}
GROUP BY facility
ORDER BY nb_arrangements DESC
""", conn))
```

```python
# SECTION 8 : Générer et télécharger le FICHIER SQL (résultat final)
# Utilité : iterdump() écrit le script SQL complet (CREATE TABLE + tous les INSERT) dans
# "imf_facilities.sql". Le fichier est enregistré dans ton Drive et téléchargé sur ton ordinateur.
sql_path = os.path.join(drive_data_dir, 'imf_facilities.sql')

with open(sql_path, "w", encoding="utf-8") as f:
    for ligne in conn.iterdump():
        f.write(ligne + "\n")

print("✅ Fichier SQL créé :", sql_path)
files.download(sql_path)
```

```python
# SECTION 9 (optionnelle) : Requêtes INSERT, UPDATE, DELETE
# Utilité : montre les opérations CRUD du tutoriel. On ajoute une ligne de test, on la modifie,
# puis on la supprime : la base revient donc exactement à son état d'origine (le fichier SQL
# de la section 8 n'est pas touché).
cursor.execute(f"""
INSERT INTO {table_name} (facility, date_of_arrangement, expiration_date, amount_agreed, amount_drawn, amount_outstanding)
VALUES ('Test Facility', '2026-01-01', '2027-01-01', 1000, 0, 0)
""")
conn.commit()
print("INSERT : ligne de test ajoutée")

cursor.execute(f"UPDATE {table_name} SET amount_drawn = 500, amount_outstanding = 500 WHERE facility = 'Test Facility'")
conn.commit()
print(f"UPDATE : {cursor.rowcount} ligne modifiée")
display(pd.read_sql_query(f"SELECT * FROM {table_name} WHERE facility = 'Test Facility'", conn))

cursor.execute(f"DELETE FROM {table_name} WHERE facility = 'Test Facility'")
conn.commit()
print(f"DELETE : {cursor.rowcount} ligne supprimée")

cursor.execute(f"SELECT COUNT(*) FROM {table_name}")
print("Lignes restantes :", cursor.fetchone()[0])
conn.close()
```

### **Question 2 : Montant moyen accordé par type de facilité**

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('/content/imf_facilities.db')

q2 = """
SELECT facility, ROUND(AVG(amount_agreed), 2) AS average_amount_agreed
FROM facilities
GROUP BY facility
ORDER BY average_amount_agreed DESC;
"""
display(pd.read_sql_query(q2, conn))
conn.close()
```

### **Question 3 : Taux de tirage par type de facilité**

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('/content/imf_facilities.db')

q3 = """
SELECT facility,
       ROUND(SUM(amount_drawn), 2) AS total_drawn,
       ROUND(SUM(amount_agreed), 2) AS total_agreed,
       ROUND((SUM(amount_drawn) / SUM(amount_agreed)) * 100, 2) || '%' AS draw_rate_pct
FROM facilities
GROUP BY facility;
"""
display(pd.read_sql_query(q3, conn))
conn.close()
```

### **Question 5 : Statistiques du montant accordé par type de facilité**

### **Question 7 : Arrangements de courte durée (< 1 an)**

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('/content/imf_facilities.db')

q7 = """
SELECT facility, date_of_arrangement, expiration_date,
       CAST((julianday(expiration_date) - julianday(date_of_arrangement)) AS INT) AS duration_days
FROM facilities
WHERE (julianday(expiration_date) - julianday(date_of_arrangement)) < 365
ORDER BY duration_days;
"""
display(pd.read_sql_query(q7, conn))
conn.close()
```

### **Question 8 : Taux de tirage par type de facilité et par décennie**

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('/content/imf_facilities.db')

q8 = """
SELECT facility,
       (strftime('%Y', date_of_arrangement) / 10) * 10 AS decade,
       ROUND(SUM(amount_drawn), 2) AS total_drawn,
       ROUND(SUM(amount_agreed), 2) AS total_agreed,
       ROUND((SUM(amount_drawn) / SUM(amount_agreed)) * 100, 2) || '%' AS draw_rate_pct
FROM facilities
GROUP BY facility, decade
ORDER BY decade ASC, facility;
"""
display(pd.read_sql_query(q8, conn))
conn.close()
```
